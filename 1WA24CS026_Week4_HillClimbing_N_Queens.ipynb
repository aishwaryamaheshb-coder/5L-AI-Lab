```python
import random
from itertools import combinations
def cost(state, show=False):
    n = len(state)
    conflicts = 0
    for i, j in combinations(range(n), 2):
        same_row = state[i] == state[j]
        same_diag = abs(i - j) == abs(state[i] - state[j])
        if same_row or same_diag:
            conflicts += 1
            if show:
                print(f"      ({i},{state[i]}) & ({j},{state[j]}) -> conflict #{conflicts}")
    return conflicts
def neighbours(state):
    result = []
    for i, j in combinations(range(len(state)), 2):
        nb = list(state)
        nb[i], nb[j] = nb[j], nb[i]
        result.append(((i, j), nb))
    return result
def print_board(state):
    n = len(state)
    for r in range(n):
        print("   " + " ".join("Q" if state[c] == r else "." for c in range(n)))
def hill_climbing(state, verbose=True):
    current = list(state)
    current_cost = cost(current)
    step = 0
    while True:
        if verbose:
            print(f"\nStep {step}: current = {current}  cost = {current_cost}")
            cost(current, show=True)
            print("   Neighbours:")
        best, best_cost, best_swap = None, None, None
        for swap, nb in neighbours(current):
            c = cost(nb)
            if verbose:
                print(f"     swap x{swap[0]},x{swap[1]} -> {nb}  cost = {c}")
            if best_cost is None or c < best_cost:
                best, best_cost, best_swap = nb, c, swap
        if best_cost >= current_cost:
            if verbose:
                print(f"\nNo neighbour better than {current_cost}. Stop.")
            return current, current_cost
        if verbose:
            print(f"   -> choose swap x{best_swap[0]},x{best_swap[1]} "
                  f"({best}, cost {best_cost})")
        current, current_cost = best, best_cost
        step += 1
def hill_climbing_random_restart(n, max_restarts=1000):
    for attempt in range(1, max_restarts + 1):
        start = random.sample(range(n), n)
        state, c = hill_climbing(start, verbose=False)
        if c == 0:
            return state, attempt
    return None, max_restarts
if __name__ == "__main__":
    print("=" * 55)
    print("4-QUEENS: initial state from the lab sheet")
    print("=" * 55)
    initial = [3, 1, 2, 0]
    final, final_cost = hill_climbing(initial)
    print("\nFinal state:", final, " cost =", final_cost)
    print_board(final)
    print("GOAL REACHED" if final_cost == 0 else "Stuck in a local minimum")

    print("\n" + "=" * 55)
    print("N-QUEENS with random restarts")
    print("=" * 55)
    random.seed(1)
    for n in (4, 8, 12):
        sol, tries = hill_climbing_random_restart(n)
        print(f"\nN = {n}: solved after {tries} restart(s) -> {sol}")
        if n <= 8:
            print_board(sol)
```
