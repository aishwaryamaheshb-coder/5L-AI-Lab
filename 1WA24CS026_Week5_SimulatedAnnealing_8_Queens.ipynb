```python
import math
import random
from itertools import combinations

def cost(state):
    return sum(1 for i, j in combinations(range(len(state)), 2)
               if state[i] == state[j] or abs(i - j) == abs(state[i] - state[j]))

def random_neighbour(state):
    nb = list(state)
    i, j = random.sample(range(len(nb)), 2)
    nb[i], nb[j] = nb[j], nb[i]
    return nb

def print_board(state):
    n = len(state)
    for r in range(n):
        print("   " + " ".join("Q" if state[c] == r else "." for c in range(n)))

def simulated_annealing(n=8, T=100.0, alpha=0.99, T_min=1e-3, verbose=True):
    current = random.sample(range(n), n)
    current_cost = cost(current)
    iteration = 0
    if verbose:
        print(f"Initial state {current}  cost = {current_cost}")
        print(f"{'iter':>5} {'T':>9} {'next cost':>9} {'dE':>4} {'accepted':>9}")
    while T > T_min and current_cost > 0:
        nxt = random_neighbour(current)
        next_cost = cost(nxt)
        delta_e = current_cost - next_cost
        if delta_e > 0:
            accepted = True
        else:
            p = math.exp(delta_e / T)
            accepted = random.random() < p
        if accepted:
            current, current_cost = nxt, next_cost
        if verbose and (iteration < 10 or iteration % 50 == 0 or current_cost == 0):
            print(f"{iteration:>5} {T:>9.4f} {next_cost:>9} {delta_e:>4} "
                  f"{'yes' if accepted else 'no':>9}   (current cost {current_cost})")
        T *= alpha
        iteration += 1
    return current, current_cost, iteration


if __name__ == "__main__":
    random.seed(7)
    state, c, iters = simulated_annealing()
    print(f"\nFinal state: {state}   cost = {c}   iterations = {iters}")
    print_board(state)
    print("GOAL REACHED" if c == 0 else "Did not reach cost 0 (try another run)")
    random.seed(0)
    runs = 200
    wins = sum(simulated_annealing(verbose=False)[1] == 0 for _ in range(runs))
    print(f"\nSuccess rate over {runs} runs: {wins}/{runs}")
```
