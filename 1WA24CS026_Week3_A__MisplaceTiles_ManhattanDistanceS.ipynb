```python
import heapq
import itertools

INITIAL = (2, 8, 3,
           1, 6, 4,
           7, 0, 5)
GOAL = (1, 2, 3,
        8, 0, 4,
        7, 6, 5)

#Heuristics
def misplaced_tiles(state, goal=GOAL):
    return sum(1 for i in range(9) if state[i] != 0 and state[i] != goal[i])


def manhattan_distance(state, goal=GOAL):
    pos = {tile: i for i, tile in enumerate(goal)}
    total = 0
    for i, tile in enumerate(state):
        if tile != 0:
            gi = pos[tile]
            total += abs(i // 3 - gi // 3) + abs(i % 3 - gi % 3)
    return total


def get_neighbors(state):
    b = state.index(0)
    r, c = divmod(b, 3)
    moves = {"Up": (-1, 0), "Down": (1, 0), "Left": (0, -1), "Right": (0, 1)}
    result = []
    for name, (dr, dc) in moves.items():
        nr, nc = r + dr, c + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            n = nr * 3 + nc
            new = list(state)
            new[b], new[n] = new[n], new[b]
            result.append((name, tuple(new)))
    return result


def inversions(state):
    tiles = [t for t in state if t != 0]
    return sum(1 for i in range(8) for j in range(i + 1, 8) if tiles[i] > tiles[j])


def is_solvable(start, goal=GOAL):
    return inversions(start) % 2 == inversions(goal) % 2


def print_state(state, indent=""):
    for r in range(3):
        print(indent + " ".join(str(x) if x else "_" for x in state[3 * r:3 * r + 3]))

def astar(start, goal, h, verbose=True):
    if not is_solvable(start):
        print("This puzzle is not solvable.")
        return None

    counter = itertools.count()
    g0 = 0
    open_list = [(g0 + h(start), next(counter), g0, start, None, None)]
    best_g = {start: 0}
    parent = {}
    closed = set()
    expanded = 0

    while open_list:
        f, _, g, state, par, move = heapq.heappop(open_list)

        if state in closed:
            continue
        closed.add(state)
        parent[state] = (par, move)
        expanded += 1

        if verbose:
            print(f"\nExpanding #{expanded}:  g={g}  h={h(state)}  f={f}")
            print_state(state, "   ")

        if state == goal:
            path = []
            while state is not None:
                p, m = parent[state]
                path.append((m, state))
                state = p
            path.reverse()
            return path, expanded

        for m, nxt in get_neighbors(state):
            ng = g + 1
            if nxt in closed:
                continue
            if nxt not in best_g or ng < best_g[nxt]:
                best_g[nxt] = ng
                heapq.heappush(open_list,
                               (ng + h(nxt), next(counter), ng, nxt, state, m))
    return None


def run_case(title, h):
    print("=" * 50)
    print(title)
    print("=" * 50)
    result = astar(INITIAL, GOAL, h)
    if result:
        path, expanded = result
        print("\n--- SOLUTION PATH ---")
        for step, (move, st) in enumerate(path):
            print(f"Step {step}: {move or 'Start'}   "
                  f"(g={step}, h={h(st)}, f={step + h(st)})")
            print_state(st, "   ")
        print(f"\nMoves in solution : {len(path) - 1}")
        print(f"Nodes expanded    : {expanded}\n")


if __name__ == "__main__":
    run_case("CASE 1: h(n) = Number of misplaced tiles", misplaced_tiles)
    run_case("CASE 2: h(n) = Manhattan distance", manhattan_distance)
```
