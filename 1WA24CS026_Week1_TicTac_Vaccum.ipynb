```python
#tictactoe
import random
M = [" ", " ", " ",
     " ", " ", " ",
     " ", " ", " "]
def check(player):
    if M[0] == M[1] == M[2] == player or M[3] == M[4] == M[5] == player or M[6] == M[7] == M[8] == player:
        return True
    if M[0] == M[3] == M[6] == player or M[1] == M[4] == M[7] == player or M[2] == M[5] == M[8] == player:
        return True
    if M[0] == M[4] == M[8] == player or M[2] == M[4] == M[6] == player:
        return True
    return False
def placex():
    while True:
        x = int(input("Enter cell (0-8): "))
        if x < 0 or x > 8:
            print("Enter a number between 0 and 8.")
        elif M[x] != " ":
            print("Cell already occupied.")
        else:
            M[x] = "X"
            break
def placeo():
    empty = []
    for i in range(9):
        if M[i] == " ":
            empty.append(i)
    x = random.choice(empty)
    M[x] = "O"
    print("Computer occupied cell:", x)
def tictactoe():
    for i in range(9):
        if i % 2 == 0:
            placex()
            if check("X"):
                print("You won!")
                return
        else:
            placeo()

            if check("O"):
                print("Computer won!")
                return
    print("It's a draw!")
tictactoe()
```

```python
#vacuumcleaner
def vacuum_cleaner():
    room_A = input("Is Room A dirty? (yes/no): ").lower()
    room_B = input("Is Room B dirty? (yes/no): ").lower()
    current_room = input("Enter the vacuum cleaner's starting room (A/B): ").upper()
    print("\nInitial State:")
    print("Room A:", room_A)
    print("Room B:", room_B)
    print("Vacuum Cleaner is in Room", current_room)
    while room_A == "yes" or room_B == "yes":
        if current_room == "A":
            if room_A == "yes":
                print("\nVacuum cleaner is in Room A.")
                print("Room A is dirty.")
                print("Cleaning Room A...")
                room_A = "no"
                print("Room A is now clean.")
            print("Moving from Room A to Room B...")
            current_room = "B"
        else:
            if room_B == "yes":
                print("\nVacuum cleaner is in Room B.")
                print("Room B is dirty.")
                print("Cleaning Room B...")
                room_B = "no"
                print("Room B is now clean.")
            print("Moving from Room B to Room A...")
            current_room = "A"
    print("\n==============================")
    print("Both rooms are clean!")
    print("Goal state reached.")
    print("==============================")
vacuum_cleaner()
```
